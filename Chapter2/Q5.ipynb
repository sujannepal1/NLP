# Exercise 5: Implement Laplace Smoothing

## Objective
Calculate word probabilities using Laplace smoothing.

```python
word_counts = {
"cat":4,
"dog":3,
"bird":1
}
```

```python
def laplace_smoothing(count , total , total_vocab, smoothing_parameter=1)-> float:
    probability = (count + smoothing_parameter) / (total + total_vocab)
    return probability
```

```python
total_words = sum(word_counts.values())
total_vocab = len(word_counts)
```

```python
laplace_smoothed_values = {f"{key}": laplace_smoothing(value,total_words,total_vocab) for key,value in word_counts.items()}
print(laplace_smoothed_values)
```

```python
sum = 0
for val in laplace_smoothed_values.values():
    sum += val

print(sum)
assert sum == 1
```
