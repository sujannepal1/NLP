# Exercise 8: Dense Vector Similarity

## Objective
1) Compute cosine similarity between two dense vectors.

```python
cat = [0.4,0.2,0.8]
dog = [0.5,0.1,0.7]
```

```python
def get_dot_product(vector1, vector2):
    dot_product = 0
    for i in range(len(vector1)):
        dot_product += vector1[i] * vector2[i]
    return dot_product

dot_product = get_dot_product(cat, dog)
print(dot_product)
```

```python
def get_denominator(vector):
    denominator = 0
    for i in range(len(vector)):
        denominator += vector[i] ** 2
    return denominator ** 0.5

denominator_cat = get_denominator(cat)
denominator_dog = get_denominator(dog)
print(denominator_cat)
print(denominator_dog)
```

```python
def cosine_similarity(a , denominator_a, denominator_b):
    return a / (denominator_a * denominator_b)

cosine_similarity_value = cosine_similarity(dot_product, denominator_cat, denominator_dog)
print(cosine_similarity_value)
```
