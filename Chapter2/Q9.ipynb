# Exercise 9: Find Similar Words
## Objective
1) Find the word most similar to a given word.

```python
embeddings = {
"cat":[0.4,0.2,0.8],
"dog":[0.5,0.1,0.7],
"car":[-0.7,0.8,-0.2],
"lion":[0.45,0.25,0.82]
}
```

```python
def get_dot_product(vec1, vec2):
    return sum(a * b for a, b in zip(vec1, vec2))

def cosine_similarity(vec1, vec2):
    dot_product = get_dot_product(vec1, vec2)
    magnitude_vec1 = sum([x**2 for x in vec1]) ** 0.5
    magnitude_vec2 = sum([x**2 for x in vec2]) ** 0.5
    if magnitude_vec1 == 0 or magnitude_vec2 == 0:
        return 0
    return dot_product / (magnitude_vec1 * magnitude_vec2)

```

```python
max_similarity = -1
max_pair = ("", "")
for i in range(len(embeddings)):
    for j in range(i + 1, len(embeddings)):
        word1 = list(embeddings.keys())[i]
        word2 = list(embeddings.keys())[j]
        similarity = cosine_similarity(embeddings[word1], embeddings[word2])
        max_similarity = max(max_similarity, similarity)
        if similarity == max_similarity:
            max_pair = (word1, word2)
        print(f"Cosine similarity between '{word1}' and '{word2}': {similarity:.4f}")

print(f"\nMaximum cosine similarity: {max_similarity:.4f}")
print(f"Pair of words with maximum similarity: {max_pair}")
```
