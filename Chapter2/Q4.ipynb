# Exercise 4: Detect Data Sparsity

## Objective

Find word pairs that never occur together.

```python

class VocabularyCounter:
    def __init__(self, value) -> None:
        self.val: Any = value
        self.vocabulary_set = set()

    def process_data_type(self, val=None):
        value = self.val if val is None else val

        if isinstance(value, str):
            self.vocabulary_set.update(value.split())

        elif isinstance(value, list):
            for item in value:
                self.process_data_type(item)

        else:
            raise NotImplementedError(f"{type(value)} not supported")

    def length(self):
        return len(self.vocabulary_set)
sentence = "the cat sat on the mat"

vocab_generator = VocabularyCounter(sentence)
vocab_generator.process_data_type()
print(vocab_generator.vocabulary_set)

length_of_vocab = vocab_generator.length()
ordered_vocabulary = sorted(vocab_generator.vocabulary_set)
print(length_of_vocab)
print(ordered_vocabulary)
vocab_dictionary = {word: index for index, word in enumerate(ordered_vocabulary)}
print(vocab_dictionary)
```

```python
import numpy as np

matrix = np.zeros((length_of_vocab, length_of_vocab))

print(matrix)
```

```python
sentence_split = sentence.split()
sentence_length = len(sentence_split)
sentence_encoding = list(enumerate(sentence.split()))
sentence_dictionary= dict(enumerate(sentence.split()))
print(sentence_dictionary)
print(sentence_encoding)
```

```python
for index, value in sentence_encoding:
    window_size = 1
    starting_index = max(index - window_size, 0)
    stopping_index = min(index + window_size, sentence_length - 1)
    print(f"index: {index}, value: {value}, starting_index: {starting_index}, stopping_index: {stopping_index}")
    index_of_current_word = vocab_dictionary[value]
    for i in range(starting_index, stopping_index):
        if i != index:
            matrix[index_of_current_word,list(vocab_dictionary.keys()).index(sentence_dictionary[i])] += 1
            matrix[list(vocab_dictionary.keys()).index(sentence_dictionary[i]),index_of_current_word] += 1

   
print(matrix)
```

```python
import pandas as pd

df = pd.DataFrame(matrix, index=ordered_vocabulary, columns=ordered_vocabulary, dtype=int)
```

# Main Code for finding words that don't occur together

```python
unique_order_of_set = set()
for i in range(length_of_vocab):
    for j in range(length_of_vocab):
        if matrix[i][j] == 0:
            val_a, val_b = sorted([i , j])
            unique_order_of_set.add((val_a, val_b))

print(unique_order_of_set)

for tups in unique_order_of_set:
    i, j = tups
    print(f"Word '{ordered_vocabulary[i]}' doesn't co-occurs with word {ordered_vocabulary[j]}")
```
