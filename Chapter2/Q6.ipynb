# Exercise 6: Simple Keyboard Prediction

## Objective
Predict the most common next word.

```python
corpus = [  
"I like NLP",
"I like Python",
"I like coffee",
"I love NLP",
"I love taste of coffee",
"I love taste of coffee",
"I love taste of coffee",
]
```

• Build bi-gram counts.
• Store next-word frequencies.
• Display the top three predictions.

```python
bi_gram_list = []
for value in corpus:
    words = value.split()
    bi_gram = bi_gram_list.extend(zip(words[:-1], words[1:]))

print(bi_gram_list)
```

```python
from collections import Counter
#frequency of the individual bi_grams
def get_count(val: list):
    freq = Counter(val).most_common()
    freq_dict = dict(freq)
    return freq_dict
```

```python
frequency = get_count(bi_gram_list)
print(frequency)
```

```python
def get_last_word_from_text(text) -> str:
        return text.split()[-1]


def predict_next_word(bi_gram_frequency_dict, text: str):
    predicted_words: list[tuple] = []
    last_word: str = get_last_word_from_text(text)
    for bi_gram, count in bi_gram_frequency_dict.items():
          print(bi_gram)
          first_word , second_word = bi_gram
          if last_word == first_word:
                predicted_words.append((second_word, count))

    print(predicted_words)
    return predicted_words

words_list: list[tuple] = predict_next_word(frequency,"I love")
print(f"words: {words_list}")
```

```python
# now get the top 3 words from the predicted words list based on their frequency.
predicted_word = []

for i in range(3):
    try:
        predicted_word.append(words_list[i][0])
    except IndexError:
        break

print(f"Predicted words: {predicted_word}")
```
