# Exercise 7: Calculate TF-IDF

## Compute TF, IDF, and TF-IDF scores.

```python
document1 = "I like NLP"
document2 = "I like AI"
document3 = "AI is amazing"
document4 = "I is amazing bruh"


document_collection = [document1, document2, document3, document4]
doc_length = len(document_collection)
```

```python
def process_document(documents:list):
    # check if document is empty
    if not documents:
        return

    tf_of_documents = []
    
    for document in documents:
        unique_word_counter = {}
        total_words = 0
        word_list = document.split()
        total_words += len(word_list)
        for word in word_list:
            if word in unique_word_counter:
                unique_word_counter[word] +=1
            else: 
                unique_word_counter[word] = 1

        print(unique_word_counter)

        term_frequency={}
        if unique_word_counter:
            term_frequency = {key : value/total_words for key,value in unique_word_counter.items()}

        print(term_frequency)
        tf_of_documents.append(term_frequency)

    return tf_of_documents

tfs = process_document(document_collection)
print(tfs)
```
