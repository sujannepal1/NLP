```python
# 2.3   A Universal Part-of-Speech Tagset
# https://www.nltk.org/book/ch05.html


# Tag	Meaning	English Examples
# ADJ	adjective	new, good, high, special, big, local
# ADP	adposition	on, of, at, with, by, into, under
# ADV	adverb	really, already, still, early, now
# CONJ	conjunction	and, or, but, if, while, although
# DET	determiner, article	the, a, some, most, every, no, which
# NOUN	noun	year, home, costs, time, Africa
# NUM	numeral	twenty-four, fourth, 1991, 14:24
# PRT	particle	at, on, out, over per, that, up, with
# PRON	pronoun	he, their, her, its, my, I, us
# VERB	verb	is, say, told, given, playing, would
# .	punctuation marks	. , ; !
# X	other	ersatz, esprit, dunno, gr8, univeristy
```

```python
def pos_tag_mapping(tag):
        tag_map = {
                'NN': 'Noun',
                'VB': 'Verb',
                'VBD': 'Verb (past tense)',
                'MD': 'Modals',
                'VBN': 'Verb (past participle)',
                'JJ': 'Adjective',
                'RB': 'Adverb',
                'PRP': 'Pronoun',
                'VBZ': 'Verb (3rd person singular present)',
                'VBG': 'Verb (gerund/present participle)',
                'DT': 'Determiner',
                '.': 'Punctuation',
                'NNS': 'Noun (plural)',
                'NN$': 'Noun (possessive)',
                'CD': 'Cardinal number',
            }
        return tag_map.get(tag, 'Other')
```

```python
with open("SujanNepal_POS_Tagging.txt", "w") as val:
    val.write(f"{'Words':20}{'POS Tags':20} {'Expanded Value':20}\n")
    val.write("-" * 60 + "\n")
    for all in pos_tags:
        if all[1].startswith('VB'):
            val.write(f"{all[0]:20} {all[1]} {pos_tag_mapping(all[1])} \n")
    val.write("\n \n" +"-" * 60 + "\n")
    val.write("Nouns \n")
    val.write("-" * 60 + "\n")
    for all in pos_tags:
            if all[1].startswith('NN'):
                val.write(f"{all[0]:20}{all[1]} {pos_tag_mapping(all[1])} \n")
```
