```python
from gensim.models import FastText
from gensim.test.utils import common_texts
```

```python
corpus = common_texts
```

```python
corpus
```

```python
model = FastText(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1
)
```

```python
word_embedding = model.wv["computer"]
print(word_embedding)
```

```python
similar_words = model.wv.most_similar(
    "computer"
)

print(similar_words)
```
