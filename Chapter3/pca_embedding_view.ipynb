```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# 1. Sample Word Vectors (300D synthetic data for illustration)
words = ["king", "queen", "man", "woman", "apple", "banana", "cat", "dog"]
np.random.seed(42)
# Creating pseudo-embeddings grouped by domain
embeddings = np.vstack([
    np.random.normal(loc=1.0, scale=0.2, size=(4, 300)), # People/Royalty
    np.random.normal(loc=-1.0, scale=0.2, size=(2, 300)), # Fruit
    np.random.normal(loc=0.0, scale=0.2, size=(2, 300))  # Animals
])

# 2. Dimensionality Reduction (PCA vs t-SNE)
# Use PCA for global structure preservation, t-SNE for local cluster separation
pca = PCA(n_components=2)
reduced_vectors = pca.fit_transform(embeddings)

# 3. Plotting
plt.figure(figsize=(8, 6))
plt.scatter(reduced_vectors[:, 0], reduced_vectors[:, 1], color='blue', edgecolors='k')

for i, word in enumerate(words):
    plt.annotate(word, xy=(reduced_vectors[i, 0], reduced_vectors[i, 1]), 
                 xytext=(5, 2), textcoords='offset points', fontsize=12)

plt.title("2D PCA Projection of Word Embeddings")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```
