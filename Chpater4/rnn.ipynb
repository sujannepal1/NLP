```python
import torch
import torch.nn as nn

# For reproducible results
torch.manual_seed(42)


# Custom RNN Cell
class CustomRNN(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()

        # Model parameters
        self.W_x = nn.Parameter(torch.randn(hidden_size, input_size))
        self.W_h = nn.Parameter(torch.randn(hidden_size, hidden_size))
        self.b = nn.Parameter(torch.zeros(hidden_size))

    def forward(self, x):
        """
        x shape: (T, input_size)
        """

        T, input_size = x.shape

        # Initial hidden state
        h = torch.zeros(self.W_h.shape[0])

        # Process each time step
        for t in range(T):
            x_t = x[t]

            # RNN recurrence
            h = torch.tanh(
                self.W_x @ x_t +
                self.W_h @ h +
                self.b
            )

        return h


# -------------------------
# Create input sequence
# -------------------------

T = 5
input_size = 3
hidden_size = 4

# Sequence containing 5 time steps
x = torch.randn(T, input_size)

# Target hidden representation
target = torch.randn(hidden_size)


# -------------------------
# Create model
# -------------------------

model = CustomRNN(input_size, hidden_size)


# -------------------------
# Forward pass
# -------------------------

final_hidden = model(x)

print("Final hidden state:")
print(final_hidden)


# -------------------------
# Compute scalar loss
# -------------------------

loss = torch.mean((final_hidden - target) ** 2)

print("\nLoss:")
print(loss.item())


# -------------------------
# Backpropagation
# -------------------------

loss.backward()


# -------------------------
# Print gradients
# -------------------------

print("\nParameter gradients:")

for name, parameter in model.named_parameters():
    print(f"\n{name}.grad:")
    print(parameter.grad)
```
