```python
import math


def sinusoidal_positional_encoding(seq_len, d_model):
    """Return array with shape (seq_len, d_model). where sequence length is the length
    of the sequence and d_model is the  embedding dimension of the model."""
    positional_encoding = [[0] * d_model for _ in range(seq_len)]
    for pos in range(seq_len):
        for i in range(d_model):
            if i % 2 == 0:
                encoding = pos / (10000 ** (i / d_model))
                positional_encoding[pos][i] = math.sin(encoding)
                print(f"Position {pos}, Dimension {i}: sin({encoding}) = {math.sin(encoding)}")
            else:
                encoding = pos / (10000 ** ((i - 1) / d_model))
                positional_encoding[pos][i] = math.cos(encoding)
                print(f"Position {pos}, Dimension {i}: cos({encoding}) = {math.cos(encoding)}")

    return positional_encoding

def print_positional_encoding(positional_encoding):
    """Print the positional encoding in a readable format."""
    print("\nPositional Encoding: \n")
    for _, encoding in enumerate(positional_encoding):
        print(f"{encoding}")

print_positional_encoding(sinusoidal_positional_encoding(3, 4))
    
```
