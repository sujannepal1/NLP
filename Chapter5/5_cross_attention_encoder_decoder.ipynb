```python
import numpy as np


def cross_attention(decoder_states, encoder_states, query_projection_weights,
                    key_projection_weights, value_projection_weights):
    """Return cross-attention outputs and weights."""

    query = np.matmul(decoder_states, query_projection_weights)
    key = np.matmul(encoder_states, key_projection_weights)
    value = np.matmul(encoder_states, value_projection_weights)

    key_dimension = query.shape[1]
    attention_scores = np.matmul(query, key.T) / np.sqrt(key_dimension)

    attention_scores = attention_scores - np.max(
        attention_scores, axis=1, keepdims=True
    )
    exponentiated_scores = np.exp(attention_scores)

    attention_weights = exponentiated_scores / np.sum(
        exponentiated_scores, axis=1, keepdims=True
    )

    attention_output = np.matmul(attention_weights, value)

    return attention_output, attention_weights

model_dimension = 8
key_dimension = 4
value_dimension = 4

sample_decoder_states = np.random.rand(3, model_dimension)
sample_encoder_states = np.random.rand(5, model_dimension)

sample_weight_query = np.random.rand(model_dimension, key_dimension)
sample_weight_key = np.random.rand(model_dimension, key_dimension)
sample_weight_value = np.random.rand(model_dimension, value_dimension)

cross_attention_output, cross_attention_weights = cross_attention(
    sample_decoder_states,
    sample_encoder_states,
    sample_weight_query,
    sample_weight_key,
    sample_weight_value
)

print("Cross-Attention Weights:\n", cross_attention_weights)
print("\nCross-Attention Output:\n", cross_attention_output)

print("\nOutput Shape:", cross_attention_output.shape)
print("Weights Shape:", cross_attention_weights.shape)
```
