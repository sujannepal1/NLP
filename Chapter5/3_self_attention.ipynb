```python
import numpy as np


# Step 1: Define the scaled dot-product attention calculation.
def scaled_dot_product_attention(query, key, value):
    # Step 2: Transpose key vectors so they align with query vectors.
    key_transpose = np.transpose(key)

    # Step 3: Calculate query-key similarity scores.
    dot_product = np.dot(query, key_transpose)

    # Step 4: Scale the scores using the key-vector dimension.
    dk = key.shape[-1]
    scaled_dot_product = dot_product / np.sqrt(dk)

    # Step 5: Convert scores to attention weights using softmax.
    softmax = np.exp(scaled_dot_product) / np.sum(
        np.exp(scaled_dot_product), axis=-1, keepdims=True
    )

    # Step 6: Combine values using the attention weights.
    output = np.dot(softmax, value)
    return output


# Step 7: Define self-attention using the input as the source of Q, K, and V.
def self_attention(X, W_Q, W_K, W_V):
    # Step 8: Project input X into query vectors.
    query = np.dot(X, W_Q)

    # Step 9: Project input X into key vectors.
    key = np.dot(X, W_K)

    # Step 10: Project input X into value vectors.
    value = np.dot(X, W_V)

    # Step 11: Calculate the attention output from Q, K, and V.
    attention_output = scaled_dot_product_attention(query, key, value)
    return attention_output


# Step 12: Define projection matrices for query, key, and value.
sample_weight_query = np.array([[1, 0], [0, 1]])
sample_weight_key = np.array([[0, 1], [1, 0]])
sample_weight_value = np.array([[2, 0], [0, 1]])

# Step 13: Create sample input embeddings.
sample_X = np.array([[1, 2], [0, 1]])

# Step 14: Run self-attention on the sample input.
self_attention_output = self_attention(
    sample_X,
    sample_weight_query,
    sample_weight_key,
    sample_weight_value,
)

# Step 15: Print the resulting context-aware output vectors.
print(f"Attention Output: {self_attention_output} \n")
```
