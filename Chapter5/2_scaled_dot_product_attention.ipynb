```python
import numpy as np


# Step 1: Define scaled dot-product attention.
def scaled_dot_product_attention(query, key, value):
    # Step 2: Transpose the key so matrix multiplication aligns query and key dimensions.
    key_transpose = np.transpose(key)

    # Step 3: Calculate similarity scores by multiplying each query by each key.
    dot_product = np.dot(query, key_transpose)

    # Step 4: Get the key dimension, which is used for scaling.
    dk = key.shape[-1]

    # Step 5: Scale the scores to reduce very large dot-product values.
    scaled_dot_product = dot_product / np.sqrt(dk)

    # Step 6: Convert scores into probabilities with softmax across each row.
    softmax = np.exp(scaled_dot_product) / np.sum(
        np.exp(scaled_dot_product), axis=-1, keepdims=True
    )

    # Step 7: Multiply the attention probabilities by the values.
    output = np.dot(softmax, value)

    # Step 8: Return the weighted value vectors.
    return output


# Step 9: Create a small query matrix containing two query vectors.
sample_query = np.array([[1, 0], [0, 1]])

# Step 10: Create a key matrix for matching against the queries.
sample_key = np.array([[1, 0], [0, 1]])

# Step 11: Create a value matrix whose values will be combined by the attention weights.
sample_value = np.array([[1, 0], [0, 1]])

# Step 12: Apply attention to the sample matrices.
attention_output = scaled_dot_product_attention(sample_query, sample_key, sample_value)

# Step 13: Print the final attention output.
print(f"Attention Output: {attention_output} \n")
```
