```python
import numpy as np


def scaled_dot_product_attention(query, key, value):
    key_transpose = np.transpose(key)
    dot_product = np.dot(query, key_transpose)
    dk = key.shape[-1]
    scaled_dot_product = dot_product / np.sqrt(dk)
    softmax = np.exp(scaled_dot_product) / np.sum(np.exp(scaled_dot_product), axis=-1, keepdims=True)
    output = np.dot(softmax, value)
    return output

sample_query = np.array([[1, 0], [0, 1]])
sample_key = np.array([[1, 0], [0, 1]])
sample_value = np.array([[1, 0], [0, 1]])
attention_output = scaled_dot_product_attention(sample_query, sample_key, sample_value)
print(f"Attention Output: {attention_output} \n")
```
