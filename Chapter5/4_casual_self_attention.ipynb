```python
import numpy as np


def causal_self_attention(query, key, value):
    """Return causally masked attention output and weights."""

    key_dimension = query.shape[1]
    attention_scores = np.matmul(query, key.T) / np.sqrt(key_dimension)

    sequence_length = query.shape[0]
    causal_mask = np.triu(np.ones((sequence_length, sequence_length)), k=1)

    attention_scores = np.where(causal_mask == 1, -np.inf, attention_scores)

    exponentiated_scores = np.exp(attention_scores)
    attention_weights = exponentiated_scores / np.sum(
        exponentiated_scores, axis=1, keepdims=True
    )
    attention_output = np.matmul(attention_weights, value)

    return attention_output, attention_weights

np.random.seed(42)

sequence_length = 5
key_dimension = 4
value_dimension = 4

sample_query = np.random.rand(sequence_length, key_dimension)
sample_key = np.random.rand(sequence_length, key_dimension)
sample_value = np.random.rand(sequence_length, value_dimension)

causal_attention_output, causal_attention_weights = causal_self_attention(
    sample_query, sample_key, sample_value
)


print("Causal Attention Weights:\n", causal_attention_weights)
print("\nAttention Output:\n", causal_attention_output)

print("\nOutput Shape:", causal_attention_output.shape)
print("Weights Shape:", causal_attention_weights.shape)

# sum of rows of weights should be 1
print("\nRow sums:", np.sum(causal_attention_weights, axis=1))
```
