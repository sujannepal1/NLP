```python
import numpy as np


# Step 1: Define a numerically stable softmax function.
def stable_softmax(values, axis=-1):
    """Numerically stable NumPy softmax."""

    # Step 2: Subtract the maximum value along the selected axis.
    shifted_values = values - np.max(values, axis=axis, keepdims=True)

    # Step 3: Exponentiate only the shifted values, which are now at most zero.
    exponentiated_values = np.exp(shifted_values)

    # Step 4: Divide by the sum of exponentials to produce probabilities.
    return exponentiated_values / np.sum(
        exponentiated_values, axis=axis, keepdims=True
    )


# Step 5: Create logits containing large values.
sample_logits = np.array([1000, 1001, 1002])

# Step 6: Demonstrate the direct softmax calculation.
direct_exponentials = np.exp(sample_logits)
direct_softmax = direct_exponentials / np.sum(direct_exponentials)

# Step 7: Apply the stable softmax implementation.
stable_softmax_output = stable_softmax(sample_logits)

# Step 8: Compare direct and stable softmax results.
print("\nDirect Exponentiation:")
print(direct_softmax)
print("\nStable Softmax:")
print(stable_softmax_output)
print("\nSum of probabilities:", np.sum(stable_softmax_output))
```
