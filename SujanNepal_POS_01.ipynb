```python
text= ""
with open("news.txt") as val:
    text = val.read()

print(text)
```

```python
from nltk import tokenize

import nltk
nltk.download('punkt')
```

```python
from nltk import word_tokenize
```

```python
nltk.download('punkt_tab')
```

```python
tokens = word_tokenize(text)
```

```python
tokens_sentence = word_tokenize(text, preserve_line=True)
```

```python
tokens_sentence
```

```python
tokens == tokens_sentence
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stems = [stemmer.stem(word) for word in tokens]
```

```python
stems
```

```python
pos_tags = nltk.pos_tag(tokens)
print(f"POS Tags: {pos_tags}")
```

```python
for all in pos_tags:
    if all[1] == 'NN':
        print(all[0])
```

```python
for all in pos_tags:
    if all[1].startswith('NN'):
        print(all[0])
```

```python
for all in pos_tags:
    if all[1].startswith('VB'):
        print(all[0])
```

```python
with open("output.txt", "w") as val:
    val.write(f"{'Words':<15}POS Tags\n")
    val.write("-" * 25 + "\n")
    for all in pos_tags:
        if all[1].startswith('VB'):
            val.write(f"{all[0]:<15} {all[1]} \n")
    for all in pos_tags:
            if all[1].startswith('NN'):
                val.write(f"{all[0]:<15}{all[1]} \n")
```
